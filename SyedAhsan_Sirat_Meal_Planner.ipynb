# **Step 1: Scrape Recipes**
Using recipe-scrapers to scrape five recipes. Storing the scraped data in a list.


```python
from recipe_scrapers import scrape_html
from urllib.request import urlopen
#using this recipes list to store receipes
recipes = []
meal_plan=[]
# Example recipe scraping
#Using a list of url for 5 recipes
url = []
url.append( 'https://www.allrecipes.com/slow-cooker-chicken-enchilada-casserole-recipe-8788263')
url.append( 'https://www.allrecipes.com/air-fried-tossed-taquitos-recipe-8784154')
url.append( 'https://www.allrecipes.com/taquito-casserole-recipe-8673259')
url.append( 'https://www.allrecipes.com/creamy-mississippi-chicken-enchiladas-recipe-8678116')
url.append(  'https://www.allrecipes.com/air-fryer-chicken-quesadillas-recipe-8651412')
i=0
while i<5:
  html = urlopen(url[i]).read().decode("utf-8")
  scraper = scrape_html(html, org_url=url[i])
  recipes.append({
  "title": scraper.title(),
  "ingredients": scraper.ingredients(),
  "instructions": scraper.instructions()
  })
  print("Recipes loaded!")
  i+=1

```

# **Step 2: Special Recipes**
Adding a Bangladeshi style Chicken Vuna (Custom) recipe

```python
chicken_vuna_recipe = {
    'title': 'Simple Chicken Vuna',
    'ingredients': [
        '1 pound chicken, cut into pieces',
        '2 tablespoons oil',
        '1 onion, chopped',
        '1/2 teaspoon turmeric',
        '1/2 teaspoon salt'
    ],
    'instructions': (
        'Heat oil in a pan, sauté onions until soft.\n'
        'Add chicken, turmeric, and salt. Cook until tender.\n'
        'Serve with rice or roti.'
    )
}

recipes.append(chicken_vuna_recipe)
```

# **Step 3: Menu and User Interaction**
Creating Menu for user

```python
# Printing the Menu
print("Menu: Enter a number to navigate the menu")
print("""
1. View Recipes
2. Add Recipe to Meal Plan
3. View Meal Plan
4. Generate Shopping List
5. Calculate Statistics
6. Exit""")

# Getting input from user
optionSelected = int(input("Enter a number: "))


#Using a loop to navigate the menu
while(optionSelected != 6):
  if optionSelected == 1:

    print("\nAvailable Recipes:")
    for i in range(len(recipes)):
        print(f"{i+1}. {recipes[i]['title']}")

  elif optionSelected ==2:
    # View available recipes
        print("\nAvailable Recipes:")
        for i in range(len(recipes)):
            print(f"{i+1}. {recipes[i]['title']}")

        # Ask user to select a recipe to add to the meal plan
        choice = int(input("Select a number to add to your meal plan: "))

        if 1 <= choice <= len(recipes):
            selected_recipe = recipes[choice - 1]
            meal_plan.append(selected_recipe)
            print(f"{selected_recipe['title']} added to your meal plan!")
        else:
            print("Invalid choice.")

  elif optionSelected ==3:
    # View meal plan (list of recipes added)
        if  len(meal_plan)==0:
            print("\nYour meal plan is empty.")
        else:
            print("\nYour Meal Plan:")
            for i in range(len(meal_plan)):
              print(f"{i+1}. {meal_plan[i]['title']}")

            # Ask user to select a recipe to view full details
            choice = int(input("Select a number to view the full Meal Plan OR Selec 0 to go back: "))
            # Display selected recipe details
            if choice==0:
              print("Going Back...")
            elif 1 <= choice <= len(meal_plan):
                selected_meal = meal_plan[choice - 1]
                print(f"\nFull Recipe: {selected_meal['title']}")
                print("Ingredients:")
                for ingredient in selected_meal['ingredients']:
                    print(f"- {ingredient}")
                print("\nInstructions:")
                print(selected_meal['instructions'])
            else:
                print("Invalid choice. Please select a valid number.")

  elif optionSelected ==4:
    # Generate shopping list from meal plan
      shopping_list = Counter()
      for meal in meal_plan:
          for ingredient in meal["ingredients"]:
              shopping_list[ingredient] += 1

      # Display shopping list in a table format
      print("\nShopping List:")
      print(tabulate(shopping_list.items(), headers=["Ingredient", "Quantity"], tablefmt="grid"))

      #Color Shopping List
      """
      Note: I tried to use this code found in assignment but it is not supported by google colab

      from termcolor import colored
      print(colored("Shopping List", "blue"))
      for item, qty in shopping_list.items():
      print(colored(item, "green"), ":", colored(qty, "blue"))
      """
      from IPython.display import display, HTML

      print(colored("Shopping List in color format", "blue"))

      # Loop through the shopping list and print each item with colored text
      for item, qty in shopping_list.items():
          # Create colored HTML text for each item and quantity
          item_colored = f'<span style="color:green;">{item}</span>'
          qty_colored = f'<span style="color:blue;">{qty}</span>'

          # Display the colored item and quantity
          display(HTML(f'{item_colored} : {qty_colored}'))


  elif optionSelected ==5:
    ingredient_counts = [len(recipe["ingredients"]) for recipe in meal_plan]
    if ingredient_counts:
      print("Average ingredients:", statistics.mean(ingredient_counts))
      print("Median ingredients:", statistics.median(ingredient_counts))
    else:
      print("No recipes in meal plan to calculate statistics.")
  else:
    print("Select an option from (1/2/3/4/0)")
  #If-Else Ends here

  # Printing the Menu
  print('............................................')
  print("Menu: Enter a number to navigate the menu")
  print("""
  1. View Recipes
  2. Add Recipe to Meal Plan
  3. View Meal Plan
  4. Generate Shopping List
  5. Calculate Statistics
  6. Exit""")

  # Getting input from user
  optionSelected = int(input("Enter a number: "))
```
